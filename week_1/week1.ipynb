```python
# Görev 1 - Python ile tek nöron forward pass yaz, kütüphane kullanmadan. (1. video)


# f(x) = (w_1 * x_1) + (w_2 * x_2) + b  olsun 


def forward_pass(w_1, x_1, w_2, x_2, b):
    return (w_1 * x_1) + (w_2 * x_2) + b
```

```python
# görselleştirme
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4)) # Tek nöron olduğu için boyutu biraz küçülttüm
ax.axis('off')

# Düğümlerin koordinatları
inputs_x = 0.2
inputs_y = [0.7, 0.3]
neuron_x = 0.8
neuron_y = 0.5  # Tek nöron olduğu için Y ekseninde tam ortaya hizalıyoruz

# Girdi (Input) düğümleri
for i, y in enumerate(inputs_y):
    circle = plt.Circle((inputs_x, y), 0.1, color='lightblue', zorder=4)
    ax.add_artist(circle)
    ax.text(inputs_x, y, f"x_{i+1}", ha='center', va='center', fontsize=15, zorder=5)

# Tek Nöron (Çıktı) düğümü
circle = plt.Circle((neuron_x, neuron_y), 0.1, color='lightgreen', zorder=4)
ax.add_artist(circle)
ax.text(neuron_x, neuron_y, "n_1\n(Çıktı)", ha='center', va='center', fontsize=12, zorder=5)

# Oklar (Bağlantılar)
# x_1'den n_1'e 
ax.annotate("", xy=(neuron_x-0.1, neuron_y+0.02), xytext=(inputs_x+0.1, inputs_y[0]), arrowprops=dict(arrowstyle="->", lw=1.5))

# x_2'den n_1'e 
ax.annotate("", xy=(neuron_x-0.1, neuron_y-0.02), xytext=(inputs_x+0.1, inputs_y[1]), arrowprops=dict(arrowstyle="->", lw=1.5))

plt.title("2 Girdili Tek Nöron Mimarisi (Görev 1)", fontsize=16)
plt.show()
```

```python
# Görev 2 - Birden fazla nörondan oluşan küçük bir katman kur, forward pass'i buna genişlet. (1. video)

def forward_pass(x_1, w_1_1, w_1_2, b_1, x_2, w_2_1, w_2_2, b_2):
    n1 = (x_1 * w_1_1) + (x_2 * w_2_1) + b_1
    n2 = (x_1 * w_1_2) + (x_2 * w_2_2) + b_2

    return n1, n2
```

```python
# Görev 3 - Basit bir loss fonksiyonu yaz. (2. video)

def loss_fn(output, y):
    return (output - y) ** 2
```

```python
# görselleştirme
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(12, 6))
ax.axis('off')

# Düğümlerin (Nodes) koordinatları
inputs_x = 0.1
inputs_y = [0.7, 0.3]

neurons_x = 0.5
neurons_y = [0.7, 0.3]

loss_x = 0.9
loss_y = [0.7, 0.3]

total_loss_x = 1.2
total_loss_y = 0.5

# 1. Girdi (Input) düğümlerini çiz
for i, y in enumerate(inputs_y):
    circle = plt.Circle((inputs_x, y), 0.1, color='lightblue', zorder=4)
    ax.add_artist(circle)
    ax.text(inputs_x, y, f"x_{i+1}", ha='center', va='center', fontsize=15, zorder=5)

# 2. Katman (Neuron/Output) düğümlerini çiz
for i, y in enumerate(neurons_y):
    circle = plt.Circle((neurons_x, y), 0.1, color='lightgreen', zorder=4)
    ax.add_artist(circle)
    ax.text(neurons_x, y, f"n_{i+1}\n(Output)", ha='center', va='center', fontsize=12, zorder=5)

# 3. Bireysel Loss Düğümlerini çiz
for i, y in enumerate(loss_y):
    circle = plt.Circle((loss_x, y), 0.1, color='salmon', zorder=4)
    ax.add_artist(circle)
    ax.text(loss_x, y, f"Loss {i+1}\n(n_{i+1} vs y_{i+1})", ha='center', va='center', fontsize=10, zorder=5)

# 4. Total Loss Düğümünü çiz
circle = plt.Circle((total_loss_x, total_loss_y), 0.12, color='crimson', zorder=4)
ax.add_artist(circle)
ax.text(total_loss_x, total_loss_y, "Total\nLoss", ha='center', va='center', fontsize=12, color='white', fontweight='bold', zorder=5)

# --- BAĞLANTILARI ÇİZİYORUZ ---
# Girdiden Nörona
ax.annotate("", xy=(neurons_x-0.1, neurons_y[0]), xytext=(inputs_x+0.1, inputs_y[0]), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.text(0.3, 0.72, "w_11", ha='center', va='center', fontsize=12, color='red')

ax.annotate("", xy=(neurons_x-0.1, neurons_y[1]), xytext=(inputs_x+0.1, inputs_y[0]), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.text(0.25, 0.55, "w_12", ha='center', va='center', fontsize=12, color='red')

ax.annotate("", xy=(neurons_x-0.1, neurons_y[0]), xytext=(inputs_x+0.1, inputs_y[1]), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.text(0.35, 0.45, "w_21", ha='center', va='center', fontsize=12, color='blue')

ax.annotate("", xy=(neurons_x-0.1, neurons_y[1]), xytext=(inputs_x+0.1, inputs_y[1]), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.text(0.3, 0.28, "w_22", ha='center', va='center', fontsize=12, color='blue')

# Nörondan Bireysel Loss'lara (Kesikli Çizgi)
ax.annotate("", xy=(loss_x-0.1, loss_y[0]), xytext=(neurons_x+0.1, neurons_y[0]), arrowprops=dict(arrowstyle="->", lw=1.5, ls="--"))
ax.text(0.7, 0.74, "Hedef y_1\nile kıyasla", ha='center', va='center', fontsize=10)

ax.annotate("", xy=(loss_x-0.1, loss_y[1]), xytext=(neurons_x+0.1, neurons_y[1]), arrowprops=dict(arrowstyle="->", lw=1.5, ls="--"))
ax.text(0.7, 0.26, "Hedef y_2\nile kıyasla", ha='center', va='center', fontsize=10)

# Bireysel Loss'lardan Total Loss'a (Toplama Noktası)
ax.annotate("", xy=(total_loss_x-0.1, total_loss_y+0.03), xytext=(loss_x+0.1, loss_y[0]), arrowprops=dict(arrowstyle="->", lw=1.5))
ax.annotate("", xy=(total_loss_x-0.1, total_loss_y-0.03), xytext=(loss_x+0.1, loss_y[1]), arrowprops=dict(arrowstyle="->", lw=1.5))

# Biasları göster
ax.text(neurons_x, neurons_y[0] + 0.15, "+ b_1", ha='center', fontsize=12, color='purple')
ax.text(neurons_x, neurons_y[1] + 0.15, "+ b_2", ha='center', fontsize=12, color='purple')

# Grafiğin sınırlarını ayarlayalım
ax.set_xlim(0, 1.4)
ax.set_ylim(0, 1)

plt.title("A'dan Z'ye Yapay Sinir Ağı Mimarisi (İleri Besleme + Hata Hesabı)", fontsize=16)
plt.show()
```

```python
# Görev 4 - Parametreleri manuel değiştirerek loss'un nasıl değiştiğini gözlemle, loss eğrisini çiz. (2. video)
x1, x2 = 3.0, 2.0
y1, y2 = 6.0, 24.0

w1_1 = 0.5
w1_2 = 2.0
w2_1 = 4.0
w2_2 = 0.5
b1   = 1.0
b2   = 1.5
```

```python
# görselleştirme - loss
import matplotlib.pyplot as plt
import numpy as np

# Test değerleri (-5 ile +5 arası 50 adım)
test_degerleri = np.linspace(-5, 5, 50)

# Sepetlerimiz (6 tane)
loss_w1_1, loss_w1_2, loss_w2_1, loss_w2_2 = [], [], [], []
loss_b1, loss_b2 = [], []

# 1. w1_1 değişiyor (w1_1 yerine "val" koyuyoruz, diğerleri kendi değerinde kalıyor)
for val in test_degerleri:
    o1, o2 = forward_pass(x1, val, w1_2, b1, x2, w2_1, w2_2, b2)
    loss_w1_1.append(loss_fn(o1, y1) + loss_fn(o2, y2))

# 2. w1_2 değişiyor
for val in test_degerleri:
    o1, o2 = forward_pass(x1, w1_1, val, b1, x2, w2_1, w2_2, b2)
    loss_w1_2.append(loss_fn(o1, y1) + loss_fn(o2, y2))

# 3. w2_1 değişiyor
for val in test_degerleri:
    o1, o2 = forward_pass(x1, w1_1, w1_2, b1, x2, val, w2_2, b2)
    loss_w2_1.append(loss_fn(o1, y1) + loss_fn(o2, y2))

# 4. w2_2 değişiyor
for val in test_degerleri:
    o1, o2 = forward_pass(x1, w1_1, w1_2, b1, x2, w2_1, val, b2)
    loss_w2_2.append(loss_fn(o1, y1) + loss_fn(o2, y2))

# 5. b1 değişiyor
for val in test_degerleri:
    o1, o2 = forward_pass(x1, w1_1, w1_2, val, x2, w2_1, w2_2, b2)
    loss_b1.append(loss_fn(o1, y1) + loss_fn(o2, y2))

# 6. b2 değişiyor
for val in test_degerleri:
    o1, o2 = forward_pass(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, val)
    loss_b2.append(loss_fn(o1, y1) + loss_fn(o2, y2))


# --- GÖRSELLEŞTİRME (2 Satır, 3 Sütun) ---
plt.figure(figsize=(15, 8))

# 1. Satır
plt.subplot(2, 3, 1)
plt.plot(test_degerleri, loss_w1_1, color='red')
plt.title("Sadece w1_1 Değişimi")
plt.grid(True)

plt.subplot(2, 3, 2)
plt.plot(test_degerleri, loss_w1_2, color='blue')
plt.title("Sadece w1_2 Değişimi")
plt.grid(True)

plt.subplot(2, 3, 3)
plt.plot(test_degerleri, loss_w2_1, color='green')
plt.title("Sadece w2_1 Değişimi")
plt.grid(True)

# 2. Satır
plt.subplot(2, 3, 4)
plt.plot(test_degerleri, loss_w2_2, color='purple')
plt.title("Sadece w2_2 Değişimi")
plt.grid(True)

plt.subplot(2, 3, 5)
plt.plot(test_degerleri, loss_b1, color='orange')
plt.title("Sadece b1 Değişimi")
plt.grid(True)

plt.subplot(2, 3, 6)
plt.plot(test_degerleri, loss_b2, color='brown')
plt.title("Sadece b2 Değişimi")
plt.grid(True)

plt.tight_layout()
plt.show()
```

```python
# Görev 5 - Sayısal türev (numerical derivative) ile basit bir gradient descent döngüsü kur: 
# parametreyi küçük adımlarla güncelleyerek loss'u düşür. (2. ve 3. video)

# f( (x+h) - f(x) ) / h 
# f(...) -> loss function
# loss'ların türevi -> minimum loss veren bulcaz

def total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2):
    o1, o2 = forward_pass(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2)
    return loss_fn(o1, y1) + loss_fn(o2, y2)


h = 0.001 #lr gibi (matematiksel formülde h diye onu kullandım)
# mantık: w = w - (türev * hız))
#         w1_1 = w1_1 + (türev * lr)
loss_hist = []

for epoch in range(150):
    current_loss = total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2)
    if epoch % 10 == 0:
        print(f"Adım (Epoch) {epoch:2d} | Güncel Hata: {current_loss:.4f}")
    
    loss_hist.append(current_loss)

    d_w1_1 = (total_loss(x1, w1_1 + h, w1_2, b1, x2, w2_1, w2_2, b2) - total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2)) / h
    d_w1_2 = (total_loss(x1, w1_1, w1_2 + h, b1, x2, w2_1, w2_2, b2) - total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2)) / h
    d_w2_1 = (total_loss(x1, w1_1, w1_2, b1, x2, w2_1 + h, w2_2, b2) - total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2)) / h
    d_w2_2 = (total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2 + h, b2) - total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2)) / h
    d_b1 = (total_loss(x1, w1_1, w1_2, b1 + h, x2, w2_1, w2_2, b2) - total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2)) / h
    d_b2 = (total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2 + h) - total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2)) / h

    w1_1 = w1_1 - (h * d_w1_1)
    w1_2 = w1_2 - (h * d_w1_2)
    w2_1 = w2_1 - (h * d_w2_1)
    w2_2 = w2_2 - (h * d_w2_2)

    b1 = b1 - (h * d_b1)
    b2 = b2 - (h * d_b2)

son_loss = total_loss(x1, w1_1, w1_2, b1, x2, w2_1, w2_2, b2)
loss_hist.append(son_loss)
print(f"Eğitim Bitti! Ulaşılan Son Hata: {son_loss:.4f}")

# --- EĞİTİM (LOSS) GRAFİĞİ ---
plt.figure(figsize=(8, 5))
plt.plot(loss_hist, color='red', linewidth=2)
plt.title("Eğitim Sürecinde Hatanın (Loss) Düşüşü")
plt.xlabel("Adım Sayısı (Epoch)")
plt.ylabel("Toplam Hata (Total Loss)")
plt.grid(True)
plt.show()
```
